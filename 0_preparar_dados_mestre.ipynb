```python
import pandas as pd

# 1. Fazer a lista dos ficheiros que queremos juntar (por ordem cronológica ajuda, mas nós vamos ordenar depois)
ficheiros = [
    'data/P1_2122.csv',
    'data/P1_2223.csv',
    'data/P1_2324.csv',
    'data/P1_2425.csv',
    'data/P1_2526.csv'
]

# 2. Ler todos os ficheiros de uma vez e juntá-los num só DataFrame Gigante
lista_dfs = [pd.read_csv(f) for f in ficheiros]
df_master = pd.concat(lista_dfs, ignore_index=True)

# 3. O PASSO MAIS IMPORTANTE: Organizar o tempo!
# Como vamos usar a "Janela Deslizante" para ver a forma recente, se as datas estiverem baralhadas, o modelo explode.
df_master['Date'] = pd.to_datetime(df_master['Date'], dayfirst=True, errors='coerce')

# Ordenar do jogo mais antigo para o mais recente e limpar os índices
df_master = df_master.sort_values(by='Date').reset_index(drop=True)

# 4. Ver o tamanho da nossa nova "Bíblia" do futebol
print(f"Bolas rolaram: {len(df_master)} jogos carregados com sucesso!")

# 5. Exportar este ficheiro mestre para não termos de fazer isto sempre
df_master.to_csv('data/P1_Mestre_5Epocas.csv', index=False)

df_master.head()
```
